```python
import pandas as pd
import os
import matplotlib.pyplot as plt
from IPython.display import clear_output

directory = "/mnt/t2/4sepmdetectshigherupwkndrun"
SKIP_ROWS = 16 #16 for tektronix, 8? for lecroy??
NUMBER_CHANNELS = 2
#TODO: make if statement for lecroy/tektronix

accepted_files = []
rejected_files = []
skipped_files = []

menu_string = "Input 'a' to accept, 'b' to go back, r to reject"

files = list(os.listdir(directory))
print(files)

for idx, file in enumerate(files):
    clear_output(wait = True)
    path = os.path.join(directory, file)
    try:
        df = pd.read_csv(path,skiprows=SKIP_ROWS)
    except:
        continue
    t = df["TIME"]
    fig, ax = plt.subplots(1,NUMBER_CHANNELS, figsize = (5*NUMBER_CHANNELS,5))
    print(f" Accepted: {len(accepted_files)}\nRejected: {len(rejected_files)}\nSkipped: {len(skipped_files)}")
    [ax[i-1].plot(t, df[f"CH{i}"]) for i in range(1,len(ax)+1)]
    fig.suptitle(file)
    plt.tight_layout()
    plt.show()
    # except:
    #     print(Exception)
    answer = input(menu_string)
    if answer == "a":
        accepted_files.append(file)
    elif answer == "b":
        clear_output(wait = True)
        file = files[idx-1]
        path = os.path.join(directory, file)
        try:
            df = pd.read_csv(path,skiprows=16)
            t = df["TIME"]
            fig, ax = plt.subplots(1,NUMBER_CHANNELS, figsize = (5*NUMBER_CHANNELS,5))
            [ax[i-1].plot(t, df[f"CH{i}"]) for i in range(1,len(ax)+1)]
            plt.tight_layout()
            plt.show()
        except:
            continue
        answer = input(menu_string)
        if answer == "a":
            if file not in accepted_files:
                accepted_files.append(file)
            else:
                change = input("File already accepted, want to reject? y for yes, enter for no")
                if change == 'y':
                    accepted_files.remove(file)
                    rejected_files.append(file)

        elif answer == 'r':
            if file not in rejected_files:
                rejected_files.append(file)
            else:
                change = input("File already rejected, want to accept? y for yes, enter for no")
                if change == 'y':
                    accepted_files.append(file)
                    rejected_files.remove(file)
            continue
        skipped_files.append(file)
    elif answer == 'r':
        if file not in rejected_files:
            rejected_files.append(file)
        continue
    else:
        skipped_files.append(file)
```
